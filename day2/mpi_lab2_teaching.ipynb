## 1. Collective Communication

### From neighbours to a whole communicator

Point-to-point communication matches a sender and receiver. A **collective** involves the participating group of a communicator. In our intracommunicator examples, every rank calls the collective, including the root.

| Intent | Routine | Result |
|---|---|---|
| Give every rank the same input | `MPI_Bcast` | Every rank receives the root's data |
| Distribute separate chunks | `MPI_Scatter` | Each rank receives its own chunk from the root |
| Collect separate chunks | `MPI_Gather` | The root stores every rank's contribution in rank order |
| Combine values, result at root | `MPI_Reduce` | The root receives a sum, maximum, etc. |
| Combine values, result everywhere | `MPI_Allreduce` | Every rank receives the reduced result |
| Collect chunks everywhere | `MPI_Allgather` | Every rank receives all contributions |
| Exchange a distinct chunk with each peer | `MPI_Alltoall` | Every rank receives one chunk from every peer |
| Compute prefixes | `MPI_Scan` | Rank $r$ receives a reduction of ranks $0$ through $r$ |
| Wait for group participation | `MPI_Barrier` | No rank returns before all ranks have entered |

“All ranks participate” does not mean the operation is `MPI_Alltoall`: a broadcast, reduction, and barrier have different purposes.

### Broadcast, reduce, and allreduce in pictures



*The solvers broadcast the validated command-line arguments this way.*



*For the residual, every rank contributes a local sum of squares; rank 0 receives their total.*



*Use allreduce when every rank needs a global quantity—for example, to make the same convergence decision. It has the result of reduce followed by broadcast, but an implementation may use a different algorithm.*

### Participation and ordering are part of correctness

1. All ranks in the communicator must call the matching collectives in the same order.
2. Rooted operations must agree on the root and communicator. Data type signatures must match; with the basic types used here, use matching types and counts as specified by each routine.
3. Put root-only processing **around the result**, not around the collective call itself.
4. Do not treat a blocking collective as a barrier unless that is its specified purpose. A routine can return when its local completion conditions hold without every other rank having returned.

These rules prevent a rank from waiting for a different operation than its peers. See [collective correctness](https://www.mpi-forum.org/docs/mpi-4.1/mpi41-report/node172.htm).

### What is being reduced in our solver?

Each process computes a local contribution

$$q_r=\sum_{(i,j)\text{ owned by rank }r}[h^2(Au-b)_{i,j}]^2.$$

The printed residual is $R=\sqrt{\sum_r q_r}$. Sum the **squares first**, then take one square root. Summing independently computed local norms gives a different quantity.

`MPI_SUM`, `MPI_MAX`, `MPI_MIN`, and `MPI_PROD` are examples of predefined reduction operators. `MPI_MAXLOC` and `MPI_MINLOC` combine a value with a location using a compatible pair datatype. The lab uses `MPI_SUM` with `MPI_DOUBLE`.

### Gather preserves contributions; reduce combines them

```text
Rank 0: q0 ─┐
Rank 1: q1 ─┼─ MPI_Gather ─> root buffer [q0, q1, q2, q3]
Rank 2: q2 ─┤                         then a local sum and sqrt
Rank 3: q3 ─┘
```

C interface:

```c
int MPI_Gather(const void *sendbuf, int sendcount, MPI_Datatype sendtype,
               void *recvbuf, int recvcount, MPI_Datatype recvtype,
               int root, MPI_Comm comm);
```

`recvcount` is the number of elements **from each rank**, not the total buffer capacity. For one `double` per rank, the root allocates space for `cells` doubles and specifies `recvcount = 1`. The root contributes too. On nonroot ranks, the receive buffer is ignored and may be `NULL`.

### Why ordering matters for floating-point addition

Finite-precision addition is not associative: $(a+b)+c$ can differ from $a+(b+c)$. A parallel reduction may group contributions in a tree. Gathering first lets you sum in a chosen rank order for this fixed decomposition.

This does **not** guarantee identical bits across process counts, compilers, or machines: changing the decomposition changes the local sums before gathering. Use numerical tolerances when comparing correct implementations. Gathering also places storage and final summation work on the root, so it is not automatically a performance improvement.

The next cell illustrates rounding in ordinary Python floating-point arithmetic; the same issue motivates the MPI exercise.

````python
a, b, c = 1e16, -1e16, 1.0
print("(a + b) + c =", (a + b) + c)
print("a + (b + c) =", a + (b + c))
````

## 2. One-sided Communication

### Origin, target, and window

In **remote memory access (RMA)**, an **origin** process initiates a transfer to or from a **target** process's exposed memory. A **window** describes that exposed memory and the processes allowed to access it. A matching send/receive pair is not required for each transfer, but setup and synchronisation still coordinate the ranks.

| Operation | Direction | Example use |
|---|---|---|
| `MPI_Put` | Origin buffer → target window | Deliver an owned boundary row to a neighbour |
| `MPI_Get` | Target window → origin buffer | Fetch a neighbour's exposed data |
| `MPI_Accumulate` | Combine origin values into target window | Apply an operation such as sum |

```text
Rank A: origin                           Rank B: target
owned row ─────── MPI_Put ─────────────> exposed window
                                         |
                              after completion/synchronisation
                                         |
                                  copy to local ghost row
```

The roles describe a transfer, not permanent kinds of process. Our ranks can be origins and targets in the **same communication epoch**. A fence does not switch a process from one role to the other.

### Allocate memory, then expose it

`MPI_Win_create(base, size, disp_unit, info, comm, &win)` is collective over `comm`, but each rank supplies its own local memory. It does not allocate that memory. Our example first allocates it with `MPI_Alloc_mem`, then exposes it with `MPI_Win_create`.

- `base`: start of this rank's exposed memory.
- `size`: size of the exposed region **in bytes**.
- `disp_unit`: scale factor **in bytes** for target displacements.
- `win`: the resulting window handle.

With `disp_unit = sizeof(double)`, a `target_disp` of 8 selects the ninth double in the target's window. It does not mean eight bytes. In general, the target address is `target_base + target_disp * target_disp_unit` in byte arithmetic.

```c
int MPI_Put(const void *origin_addr, int origin_count,
            MPI_Datatype origin_datatype, int target_rank,
            MPI_Aint target_disp, int target_count,
            MPI_Datatype target_datatype, MPI_Win win);
```

The origin and target counts describe compatible data signatures. The target range must fit in the exposed window. See [the MPI put definition](https://www.mpi-forum.org/docs/mpi-4.1/mpi41-report/node317.htm).

### Fences delimit a safe communication interval

An **access epoch** is an interval in which an origin may issue RMA operations. An **exposure epoch** is the corresponding interval during which a target exposes its window. In the active-target fence scheme used here, ranks collectively call `MPI_Win_fence` on the window to delimit these intervals.

```text
All ranks, in the same window group:

fence ── issue puts to neighbours ── fence ── use received window data
  ^                                   ^
open the transfer interval       complete preceding RMA transfers
```

Returning from `MPI_Put` alone does not establish completion or make the target's values ready for use. The closing fence supplies the completion and visibility guarantees needed by this example. The origin must also preserve its send data until completion. The following iteration repeats the fence/transfer/fence sequence.

Use `MPI_Win_fence(0, win)` for the basic pattern. The first argument is an assertion bitmask used for possible optimisation, not a rank or a success flag. Only make stronger assertions when their conditions actually hold. [Fence synchronisation](https://www.mpi-forum.org/docs/mpi-4.1/mpi41-report/node327.htm) defines these epochs.

### Design the single-window layout before editing

The starter uses a separate window for each ghost row. The exercise combines them. Sketch a window of `2 * mesh_size` doubles and label which half receives data from which neighbour. For each outgoing row, write down the **receiver's** destination slot and derive `target_disp` in units of doubles.

```text

```

Keep concurrent puts to distinct locations. An absent neighbour must not cause uninitialised window contents to overwrite a physical boundary. In particular, a one-rank run has two physical boundaries and no incoming halo data.

After the final epoch, collectively release the window with `MPI_Win_free` and release the separately allocated memory with `MPI_Free_mem`.

The mechanism of one-sided communication (aka remote memory access) proposes an entirely different communication pattern than the message-passing examples that we have seen. An origin issues a transfer to a target window without a matching send or receive call at the target. The processes still need appropriate synchronisation before accessing the transferred data. Instead, by creating a memory object called window at the target in which the memory window is accessed, the origin, in which the call is performed, can put data to and get data from that window followed by whatever local load/store operations.

In our example [laplace_mpi_win.c](./laplace_mpi_win.c), each MPI process is used as both the origin and the target - when the window is created and exposed, it is treated as the target, when an RMA call is performed, it is the origin. 

`As the target`, two window objects are created with separate allocated memory in an MPI process, one for the top row of ghost nodes, the other for the bottom row of ghost nodes.
```cpp
MPI_Win upper_win, lower_win;
double *upper_bnd, *lower_bnd;
/* win for the top row of ghost nodes */
MPI_Alloc_mem(sizeof(double )*mesh_size, MPI_INFO_NULL, &upper_bnd);
MPI_Win_create(upper_bnd, mesh_size *sizeof(double), sizeof(double),\
               MPI_INFO_NULL, MPI_COMM_WORLD, &upper_win);

/* win for the bottom row of ghost nodes */
MPI_Alloc_mem(sizeof(double )*mesh_size, MPI_INFO_NULL, &lower_bnd);
MPI_Win_create(lower_bnd, mesh_size *sizeof(double), sizeof(double),\
               MPI_INFO_NULL, MPI_COMM_WORLD, &lower_win); 
```

`As the origin`, the MPI process is very much like the sender doing `MPI_Send`, it needs to transfer out the top and bottom full nodes for the neighbouring processes to update their ghost nodes, however, this is done by using `MPI_Put`.

```cpp
/* put the bottom row of full nodes to the rank below */
MPI_Put(submesh[1], mesh_size, MPI_DOUBLE, lower, 0, mesh_size, MPI_DOUBLE, upper_win);

/* put the top row of full nodes to the rank above */
MPI_Put(submesh[*ptr_rows - 2], mesh_size, MPI_DOUBLE, higher, 0, mesh_size, MPI_DOUBLE, lower_win);   
```

Next, after completing the RMA operation the same MPI process needs to load/store data from its window object to update its own ghost nodes. This makes it the `target`.

```cpp
/* Preserve physical boundaries; copy ghost rows only for real neighbours. */
if (lower != MPI_PROC_NULL) {
    for (int i = 0; i < mesh_size; i++) {
        submesh[0][i] = lower_bnd[i];
    }
}
if (higher != MPI_PROC_NULL) {
    for (int i = 0; i < mesh_size; i++) {
        submesh[*ptr_rows - 1][i] = upper_bnd[i];
    }
}
```

The same rank can act as origin and target during one epoch. The fences delimit the RMA transfer interval and make the received window data safe to use; they do not switch a rank's role.

A window can receive multiple puts from processes in its communicator. Choose nonconflicting target locations and complete the RMA epoch before accessing the received data locally.
After freeing each window with `MPI_Win_free`, release its allocated memory with `MPI_Free_mem`.

## 3. MPI-IO

### Why one shared output file?

Each rank owns only part of the solution. Writing one file per rank creates a later assembly task and potentially many small files. Gathering everything to one root creates a root-memory and output bottleneck. MPI-IO lets ranks address their parts of a common file directly.

MPI-IO is an API and can work with supported ordinary filesystems as well as parallel filesystems. A parallel filesystem can provide the aggregate bandwidth needed at scale; it is not a universal requirement for functional correctness. For production runs, follow your site's current filesystem and striping guidance rather than treating slide-era size thresholds as fixed rules.

### Collective file management, independent data access

| Routine | Participation in this lab | Purpose |
|---|---|---|
| `MPI_File_open` | Collective | Open the same file with consistent access modes |
| `MPI_File_set_size` | Collective | Set the intended length; remove stale trailing data from an older run |
| `MPI_File_write_at` | Independent | Write this rank's data at an explicit offset |
| `MPI_File_write_at_all` | Collective | Alternative collective write, allowing MPI to coordinate access |
| `MPI_File_close` | Collective | Close the file handle |

`MPI_MODE_CREATE | MPI_MODE_WRONLY` creates the file if absent and opens it for writing; it does not truncate an existing file. The code sets its size explicitly. Do not call collective file-management routines only on rank 0.

The write routine's C signature is:

```c
int MPI_File_write_at(MPI_File fh, MPI_Offset offset, const void *buf,
                      int count, MPI_Datatype datatype, MPI_Status *status);
```

`buf` supplies data to write. `MPI_File_read_at` is the corresponding **read** routine. This lab writes native doubles without a header; MPI-IO does not automatically format numbers as text. A program can format text itself and write character data.

### Offset units: distinguish the file from the memory buffer

`count` counts memory-buffer elements of `datatype`. An explicit file `offset` counts elementary file types (etypes) relative to the current file view. The default view uses bytes, so offsets in this lab are byte offsets. If you change the file view, revisit that assumption. See the [file-view definitions](https://www.mpi-forum.org/docs/mpi-4.1/mpi41-report/node350.htm) and [data-access routines](https://www.mpi-forum.org/docs/mpi-4.1/mpi41-report/node363.htm).

### Derive the output layout from ownership

Write each global row exactly once. Exclude internal ghost rows and include the two physical boundary rows. For a row-major grid, the byte position of global row $g$ is

$$\text{offset}(g)=g\times\text{mesh\_size}\times\operatorname{sizeof}(\text{double}).$$

Use `MPI_Offset` arithmetic when calculating file positions. The first owned global row of rank $r$ is $r\times\text{int\_rows}+1$; rank 0 additionally writes the physical row 0.

**Worked layout:** a 10×10 grid on three ranks has eight interior rows, split as 2, 2, and 4. With 8-byte doubles:

| Rank | Global rows written | Number of doubles | First byte | Byte interval |
|---|---|---:|---:|---|
| 0 | 0–2, including bottom boundary | 30 | 0 | `[0, 240)` |
| 1 | 3–4 | 20 | 240 | `[240, 400)` |
| 2 | 5–9, including top boundary | 50 | 400 | `[400, 800)` |

The intervals meet without overlap or gaps. The final file contains 100 doubles. A one-rank run writes the entire mesh, including both physical boundaries.

After the specified number of iterations, we want to write the approximation over the unit square domain for the visualisation, validation and whatnots. The question now is the following: each MPI process only possesses a subdomain so how to merge the data to a common file?

Traditionally, one may either send all the subdomains to a root process, such is called master-slave communication mode, or output each individual subdomains for a different program to scan and combine.

MPI-IO provides the ability for multiple MPI processes to write into a common file. The basic MPI-IO procedures resembles with the POSIX IO:
```cpp
/* Open a file */
MPI_File_open(MPI_COMM_WORLD, file_name, MPI_MODE_CREATE|MPI_MODE_WRONLY, MPI_INFO_NULL, &fp);

/* Resize collectively so a smaller rerun does not leave stale data. */
MPI_File_set_size(fp, (MPI_Offset)mesh_size * mesh_size * sizeof(double));

/* Write to a file */
MPI_File_write_at(fp, offset, &submesh[1][0], count, MPI_DOUBLE, &IO_status );

/* Close a file */
MPI_File_close(&fp);

```
Each process starts to write at the offset position of the common file. It is the programmer's responsibility to ensure the offset is set correctly.
With the default file view used here, offsets are measured in bytes. Only owned rows are written; internal ghost rows must be excluded. The first and last physical boundary rows must each appear once, including when a single rank owns the whole grid.

## 4. Profile with mpiP

### How PMPI enables profiling

The MPI profiling interface provides corresponding `PMPI_` entry points. A profiling library can intercept a call such as `MPI_Send`, measure around a call to `PMPI_Send`, and return its result to the application.

```text
Application calls MPI_Send
          |
Profiler wrapper starts a timer
          |
Wrapper calls PMPI_Send ──> MPI implementation performs the send
          |
Wrapper records elapsed time / call site / message information
          |
Return to the application
```

The wrapper uses `PMPI_Send` internally so it does not recursively call its own `MPI_Send` wrapper. This instrumentation observes the application's existing MPI calls; it does not change the mathematical algorithm. Instrumentation itself can add overhead.

### Read the mpiP report as evidence, not a ranking by itself

mpiP is a lightweight statistical MPI profiler built on this interface. Read the report's task summaries and call-site tables together:

| Item | What to ask |
|---|---|
| Application time | How much elapsed time is covered by the measured application interval? |
| MPI time and MPI percentage | How much time is spent inside MPI routines? |
| Call count | Is communication happening as often as the algorithm requires? |
| Bytes / message sizes | Are there many small transfers or fewer large transfers? |
| Variation between ranks | Is one rank arriving late or doing more work? |
| Dominant call sites | Which source locations account for the observed costs? |

mpiP defines application time from the end of `MPI_Init` to the beginning of `MPI_Finalize`. Report fields and aggregation matter: time accumulated over ranks is not the elapsed duration of the job. See the [mpiP user guide](https://software.llnl.gov/mpiP/).

A large time in `MPI_Waitall` may reflect transfers, delayed peers, or little independent work left to hide communication. A large time in a collective may reflect load imbalance before that call. A smaller MPI percentage alone does not establish a faster program; time may have moved elsewhere.
